```python
import pandas as pd
from collections import Counter
import matplotlib.pyplot as plt

jobs = [
    {"role": "Data Analyst Intern",
     "skills": ["Python", "SQL", "Pandas", "Excel", "Power BI", "Statistics"]},
    {"role": "Data Scientist Intern",
     "skills": ["Python", "SQL", "Pandas", "Machine Learning",
                "Scikit-learn", "Statistics"]},
    {"role": "Machine Learning Intern",
     "skills": ["Python", "Machine Learning", "Scikit-learn",
                "Pandas", "Statistics", "Git"]},
    {"role": "Python Developer Intern",
     "skills": ["Python", "SQL", "Git"]},
    {"role": "Business Intelligence Intern",
     "skills": ["SQL", "Excel", "Power BI", "Tableau"]},
    {"role": "Data Engineering Intern",
     "skills": ["Python", "SQL", "Git", "Pandas"]},
    {"role": "Research Intern",
     "skills": ["Python", "Statistics", "Machine Learning"]},
    {"role": "Data Analytics Intern",
     "skills": ["Python", "SQL", "Excel", "Pandas", "Tableau"]}
]

df_jobs = pd.DataFrame(jobs)

skill_counts = Counter(
    skill
    for skills in df_jobs["skills"]
    for skill in skills
)

skill_demand = pd.DataFrame(
    skill_counts.items(), columns=["Skill", "Count"]
).sort_values("Count", ascending=True)

print(skill_demand.sort_values("Count", ascending=False))

plt.figure(figsize=(9, 6))
plt.barh(skill_demand["Skill"], skill_demand["Count"])
plt.title("Skill Frequency Across Sample Internship Roles")
plt.xlabel("Number of roles mentioning the skill")
plt.tight_layout()
plt.savefig("skill_demand.png", dpi=150)
plt.show()
```
