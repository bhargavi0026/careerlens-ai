```python

import re
import pandas as pd

# Skills our first version can recognize
SKILLS = [
    "Python", "SQL", "Pandas", "Machine Learning",
    "Scikit-learn", "Excel", "Power BI", "Tableau",
    "Statistics", "Git"
]

# Example candidate profile
candidate_skills = ["Python", "Pandas", "SQL", "Git"]

# Sample job descriptions for testing
jobs = [
    {
        "title": "Data Analyst Intern",
        "description": (
            "Looking for a Data Analyst with Python, SQL, "
            "Pandas, Excel, Power BI, and Statistics."
        )
    },
    {
        "title": "Data Scientist Intern",
        "description": (
            "Looking for a Data Scientist with Python, Pandas, "
            "SQL, Machine Learning, Scikit-learn, and Statistics."
        )
    },
    {
        "title": "Python Developer Intern",
        "description": (
            "Looking for a Python Developer with Python, SQL, and Git."
        )
    }
]


def extract_skills(text):
    """Find recognized skills in a piece of text."""
    found = []

    for skill in SKILLS:
        pattern = r"\b" + re.escape(skill) + r"\b"

        if re.search(pattern, text, flags=re.IGNORECASE):
            found.append(skill)

    return found


def analyze_job(job, candidate_skills):
    required = extract_skills(job["description"])

    candidate_set = {s.casefold() for s in candidate_skills}
    matched = [s for s in required if s.casefold() in candidate_set]
    missing = [s for s in required if s.casefold() not in candidate_set]

    score = round(len(matched) / len(required) * 100) if required else 0

    return {
        "Job Title": job["title"],
        "Match Score (%)": score,
        "Matched Skills": ", ".join(matched) or "None",
        "Missing Skills": ", ".join(missing) or "None"
    }


results = pd.DataFrame([
    analyze_job(job, candidate_skills) for job in jobs
])

print("Candidate skills:", ", ".join(candidate_skills))
results
```
